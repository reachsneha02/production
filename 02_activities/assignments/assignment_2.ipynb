# Load the data

Place the files in the ../../05_src/data/fires/ directory and load the appropriate file. 

```python
# Load the libraries as required.
import pandas as pd
import pickle
import shap
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, PowerTransformer, OneHotEncoder
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import make_scorer, mean_squared_error


```

```python
# Load data
columns = [
    'coord_x', 'coord_y', 'month', 'day', 'ffmc', 'dmc', 'dc', 'isi', 'temp', 'rh', 'wind', 'rain', 'area' 
]
fires_dt = (pd.read_csv('../../05_src/data/fires/forestfires.csv', header = 0, names = columns))
fires_dt.info()
```

# Get X and Y

Create the features data frame and target data.

```python
num_cols_1 = ['coord_x', 'coord_y', 'ffmc', 'dmc', 'dc', 'isi', 'temp', 'wind', 'rh','rain']
num_cols_2 = ['wind', 'rain']
cat_cols = [ 'month' , 'day' ]
```

# Preprocessing

Create two [Column Transformers](https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html), called preproc1 and preproc2, with the following guidelines:

- Numerical variables

    * (Preproc 1 and 2) Scaling: use a scaling method of your choice (Standard, Robust, Min-Max). 
    * Preproc 2 only: 
        
        + Choose a transformation for any of your input variables (or several of them). Evaluate if this transformation is convenient.
        + The choice of scaler is up to you.

- Categorical variables: 
    
    * (Preproc 1 and 2) Apply [one-hot encoding](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.OneHotEncoder.html) where appropriate.


+ The only difference between preproc1 and preproc2 is the non-linear transformation of the numerical variables.
    


### Preproc 1

Create preproc1 below.

+ Numeric: scaled variables, no other transforms.
+ Categorical: one-hot encoding.

```python
preproc_1 = ColumnTransformer([
    ('numeric_simple', StandardScaler(), num_cols_1),
    ('onehot', OneHotEncoder(handle_unknown='infrequent_if_exist'), cat_cols)],
 remainder='passthrough')
```

### Preproc 2

Create preproc1 below.

+ Numeric: scaled variables, non-linear transformation to one or more variables.
+ Categorical: one-hot encoding.

```python
scal_trans = Pipeline([
    ('standardizer', StandardScaler()),
])

nonlin_trans = Pipeline([
    ('standardizer', StandardScaler()),
    ('transform', PowerTransformer(method='yeo-johnson'))
])


preproc_2 = ColumnTransformer([
    ('numeric_simple', scal_trans , list(set(num_cols_1) - set(num_cols_2))),
    ('non_linear', nonlin_trans , num_cols_2),
    ('onehot', OneHotEncoder(handle_unknown='infrequent_if_exist'), cat_cols)],
 remainder='passthrough')
```

## Model Pipeline


Create a [model pipeline](https://scikit-learn.org/stable/modules/generated/sklearn.pipeline.Pipeline.html): 

+ Add a step labelled `preprocessing` and assign the Column Transformer from the previous section.
+ Add a step labelled `regressor` and assign a regression model to it. 

## Regressor

+ Use a regression model to perform a prediction. 

    - Choose a baseline regressor, tune it (if necessary) using grid search, and evaluate it using cross-validation.
    - Choose a more advance regressor, tune it (if necessary) using grid search, and evaluate it using cross-validation.
    - Both model choices are up to you, feel free to experiment.

```python
# Pipeline A = preproc1 + baseline
PipeA = Pipeline([
    ('preprocess', preproc_1),
    ('regressor', KNeighborsRegressor(n_neighbors=5))
])
```

```python
# Pipeline B = preproc2 + baseline
PipeB = Pipeline([
    ('preprocess', preproc_2),
    ('regressor', KNeighborsRegressor(n_neighbors=5))
])
```

```python
# Pipeline C = preproc1 + advanced model
PipeC = Pipeline([
    ('preprocess', preproc_1),
    ('regressor', RandomForestRegressor(random_state=42))
])
```

```python
# Pipeline D = preproc2 + advanced model
PipeD = Pipeline([
    ('preprocess', preproc_2),
    ('regressor', RandomForestRegressor(random_state=42))
])
    
```

# Tune Hyperparams

+ Perform GridSearch on each of the four pipelines. 
+ Tune at least one hyperparameter per pipeline.
+ Experiment with at least four value combinations per pipeline.

```python

X = fires_dt.drop(columns = 'area')
Y = fires_dt['area']

scoring = {
    'rmse': make_scorer(mean_squared_error, squared=False),
    'r2': 'r2'
}


X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42)
```

```python
param_gridA = {
    'regressor__n_neighbors' : [ 3, 5, 7, 9 ] 
 }

grid_cvA = GridSearchCV(
    estimator=PipeA, 
    param_grid=param_gridA, 
    scoring=scoring, 
    cv = 5,
    refit = 'rmse')
grid_cvA.fit(X_train, Y_train)
grid_cvA.best_params_, grid_cvA.best_score_
```

```python

param_gridB = {
    'regressor__n_neighbors' : [ 3, 5, 7, 9 ] 
 }

grid_cvB = GridSearchCV(
    estimator=PipeB, 
    param_grid=param_gridB, 
    scoring=scoring, 
    cv = 5,
    refit = 'rmse')

grid_cvB.fit(X_train, Y_train)
grid_cvB.best_params_, grid_cvB.best_score_
```

```python

param_gridC = {
    'regressor__n_estimators' : [ 50, 100, 200, 300 ] 
 }

grid_cvC = GridSearchCV(
    estimator=PipeC, 
    param_grid=param_gridC, 
    scoring=scoring, 
    cv = 5,
    refit = 'rmse')

grid_cvC.fit(X_train, Y_train)
grid_cvC.best_params_, grid_cvC.best_score_
```

```python
param_gridD = {
    'regressor__n_estimators' : [ 50, 100, 200, 300 ] 
 }

grid_cvD = GridSearchCV(
    estimator=PipeD, 
    param_grid=param_gridD, 
    scoring=scoring, 
    cv = 5,
    refit = 'rmse')
grid_cvD.fit(X_train, Y_train)
grid_cvD.best_params_, grid_cvD.best_score_
```

# Evaluate

+ Which model has the best performance?

The best-performing model is PipeD, which combines a Random Forest Regressor with an enhanced preprocessing pipeline (preproc_2) that includes both standard scaling and power transformations (Non-linear transformation). With n_estimators = 50, this model achieved the lowest cross-validated RMSE of 49.57, indicating superior predictive accuracy compared to other tested models
